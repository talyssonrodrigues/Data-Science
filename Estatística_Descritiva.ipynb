## Medidas de Posição

São medidas que ajudam a entender a distribuição e a localização dos dados em relação a intervalos.

Aliás, um conceito importante aqui é o de quantis, que são os pontos que dividem um conjunto de dados em partes iguais. Estes intervalos contarão com uma certa proporção de observações abaixo ou acima deles.

👉🏼 Portanto, as medidas de posição irão lidar com diferentes tipos de quantis para ajudar a entender a distribuição e a posição dos dados dentro do conjunto.

```python
import pandas as pd

dados = {
    'Nome': [
        'Ana', 'Carlos', 'Maria', 'João', 'Lucas', 'Laura', 'Gabriel',
        'Fernanda', 'Juliana', 'Ricardo'
    ],
    'Idade': [23, 35, 23, 41, 22, 34, 26, 40, 28, 22]
}
df = pd.DataFrame(dados)
print(df)
```

### Mediana

A mediana é considerada um quantil, pois divide um conjunto de dados ordenado em duas partes iguais. Ou seja: 50% dos dados têm valores menores ou iguais à mediana e 50% deles têm valores maiores ou iguais a ela.

```python
mediana = df['Idade'].median()
print("Mediana:", mediana)
```

Nesse conjunto, significa que 50% dos datos tem valores menores ou iguais a 27

### Quartis

Dividem o conjunto de dados em 4 partes iguais:

- Q1 (1º quartil): Indica o valor abaixo do qual 25% dos dados estão localizados (= 25% dos dados têm valores menores ou iguais a Q1).

- Q2 (2º quartil): É a mediana, que divide os dados ao meio (= 50% dos dados têm valores menores ou iguais a Q2).

- Q3 (3º quartil): Indica o valor abaixo do qual 75% dos dados estão localizados (= 75% dos dados têm valores menores ou iguais a Q3).

```python
quartis = df['Idade'].quantile([0.25,0.5,0.75])
print('Quartis (Q1, Mediana, Q3)')
print(quartis)
```

25% dos valores desse conjunto estão abaixo ou igual a 23

50% abaixo ou igual a 27

75% sao menores ou iguais a 34,75

### Percentis

Dividem o conjunto de dados em 100 partes iguais.

```python
percentis = df['Idade'].quantile([0.1, 0.5, 0.9])
print("Percentis (10º, 50º e 90º):")
print(percentis)
```

10º Percentil = 22.0 → significa que 10% dos dados têm valores menores ou iguais a 22

50º Percentil = 27.0 → significa que 50% dos dados têm valores menores ou iguais a 27

90º Percentil = 40.1 → significa que 90% dos dados têm valores menores ou iguais a 40.1

### Decis

São quantis que dividem os dados em 10 partes iguais.

O 1º decil (D1), por exemplo, é o valor abaixo do qual 10% dos dados estão localizados; o 5º decil (D5) é a mediana; e o 9º decil (D9) é o valor abaixo do qual 90% dos dados se encontram.

```python
decis = df['Idade'].quantile([0.1, 0.5, 0.9])
print("Decis (D1 e D9):")
print(decis)
```

1º Decil (D1) = 22.0 (semelhante ao 10º percentil)
5º Decil (D5) = 27.0 (semelhante ao 50º percentil)
9º Decil (D9) = 40.1 (semelhante ao 90º percentil)

## Medidas de Frequência

São medidas que ajudam a entender como os dados se distribuem dentro de um conjunto, ou seja, com que frequência um valor ou grupo de valores ocorre.

São muito úteis na análise exploratória de dados, principalmente quando estamos lidando com variáveis categóricas ou discretas.

```python
import pandas as pd

data = {
    'Gênero': [
        'Masculino', 'Feminino', 'Feminino', 'Masculino', 'Feminino',
        'Feminino', 'Masculino', 'Feminino', 'Feminino', 'Feminino'
    ],
    'Faixa Etária': [
        '18-25', '26-35', '26-35', '18-25', '18-25', '36-45', '18-25',
        '26-35', '18-25', '36-45'
    ]
}

df = pd.DataFrame(data)
print(df)
```

### Frequência Absoluta

Conta o número de vezes que um determinado valor ou categoria ocorre em um conjunto de dados.

Através do método value_counts()

```python
# Frequência absoluta para a coluna 'Gênero'
frequencia_absoluta_genero = df['Gênero'].value_counts()
print("Frequência Absoluta - Gênero:")
print(frequencia_absoluta_genero)

# Frequência absoluta para a coluna 'Faixa Etária'
frequencia_absoluta_faixa_etaria = df['Faixa Etária'].value_counts()
print("\nFrequência Absoluta - Faixa Etária:")
print(frequencia_absoluta_faixa_etaria)
```

### Frequência Relativa

Refere-se à proporção de um determinado valor em um conjunto de dados em relação ao número total de observações . Ou seja, a frequência relativa de uma categoria é sua frequência absoluta (quantas vezes ela aparece) dividida pelo total de observações.

👉🏼 Essa medida ajuda a entender entender o peso relativo de cada categoria no conjunto de dados de forma simples e eficiente, especialmente para o caso das variáveis categóricas.

Através de **value_counts(normalize=True)**, sendo que o método **value_counts() **conta quantas vezes cada valor único aparece e o parâmetro **normalize(True)** retorna a proporção, ou seja, a fração de cada valor no total de registros dessa coluna.

```python
# Frequência relativa para a coluna 'Gênero'
frequencia_relativa_genero = df['Gênero'].value_counts(normalize=True)
print("\nFrequência Relativa - Gênero:")
print(frequencia_relativa_genero)

# Frequência relativa para a coluna 'Faixa Etária'
frequencia_relativa_faixa_etaria = df['Faixa Etária'].value_counts(normalize=True)
print("\nFrequência Relativa - Faixa Etária:")
print(frequencia_relativa_faixa_etaria)
```

## Medidas de Forma

São medidas que ajudam a entender a distribuição dos dados, no sentido de se a distribuição é simétrica ou assimétrica e se tem caudas mais pesadas (distribuições mais alongadas) ou mais curtas.

Essas medidas são essenciais quando queremos entender o formato de uma distribuição sem precisar necessariamente plotá-la.

### Assimetria (Skewness)

Medida que indica se a distribuição dos dados é simétrica ou se tem uma cauda mais longa para um lado. Ela pode ter três comportamentos principais:

- **Simetria:** A distribuição é aproximadamente simétrica. Se a distribuição é simétrica — como a distribuição normal — a assimetria será zero e a visualização seria algo como uma “curva de sino”.

- **Assimetria positiva**: A distribuição tem uma cauda mais longa à direita da média e a assimetria tem valores mais altos. Visualmente, a “barriga” da distribuição ficaria mais próxima do lado esquerdo e a cauda se estenderia para a direita.

- **Assimetria negativa:** A distribuição tem uma cauda mais longa à esquerda da média e a assimetria tem valores mais baixos. Visualmente, a “barriga” da distribuição ficaria mais à direita, com a cauda se estendendo para a esquerda.

👉🏼 Portanto, a assimetria nos ajuda a entender a direção da distribuição dos dados: se ela está simétrica ou inclinada.

método skew()

```python
import pandas as pd

dados = {'Funcionário': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
         'Salário': [
             2500, 3000, 2700, 3200, 2900, 18000, 4000, 3500, 2200, 2600
         ]}

df = pd.DataFrame(dados)

# Calculando a assimetria (skewness)
assimetria = df['Salário'].skew()

print(f'Assimetria (Skewness) dos salários: {assimetria}')
```

Assimetria dos salários = ~ 3.09. Esse valor positivo indica uma distribuição assimétrica positiva do conjunto de dados acima. com uma cauda à direita, certamente puxada pelo salário muito mais alto observado (18000)

Plotando o histograma

```python
import matplotlib.pyplot as plt

plt.hist(df['Salário'], bins=10, edgecolor='black', alpha=0.6, density=True)
df['Salário'].plot(kind='kde', color='red', lw=2)

plt.title('Distribuição dos Salários com Curva de Densidade')
plt.xlabel('Salário')
plt.ylabel('Densidade')
plt.grid(True)
plt.show()
```

###  Curtose

Mede a “extremidade” ou peso das caudas e como os dados estão distribuídos em torno da média.

A curtose pode ser:

**Positiva (alta curtose):**
- existem mais valores extremos;
- a distribuição pode apresentar outliers;
- alguns dados estão muito distantes do centro.

**Negativa (baixa curtose):**
- há menos valores extremos;
- os dados tendem a estar mais distribuídos;
- a distribuição pode parecer mais achatada.

**Se a curtose for próxima de zero** (ou 3 no caso da curtose excessiva), a distribuição é semelhante às distribuições normais.

método kurt()

```python
import pandas as pd

dados = {'Aluno': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
         'Nota': [50, 55, 65, 70, 80, 85, 90, 95, 100, 150]}

df = pd.DataFrame(dados)

# Calculando a curtose (kurtosis)
curtose = df['Nota'].kurt()
assimetria = df['Nota'].skew()

print(f'Curtose (Kurtosis) das notas: {curtose}')
```

Histograma

```python
plt.hist(df['Nota'], bins=10, edgecolor='black', alpha=0.6, density=True)
df['Nota'].plot(kind='kde', color='orange', lw=2)

plt.title('Distribuição das Notas com Curva de Densidade')
plt.xlabel('Nota')
plt.ylabel('Densidade')
plt.grid(True)
plt.show()
```

### Medidas de correlação

São usadas para indicar se existe uma associação linear entre as variáveis e qual o tipo dessa relação (se positiva, negativa ou nenhuma), o que é útil para modelagem preditiva, identificação de dependências e outros tipos de análises.

**Tipos comuns de medidas de correlação:**

- Coeficiente de Correlação de Pearson
- Coeficiente de Correlação de Spearman

### Coeficiente de Correlação de Pearson

- Valores próximos de 1 → Relação forte e positiva (as variáveis crescem juntas)
- Valores próximos de -1 → Relação forte e negativa (uma cresce, a outra diminui)
- Valores próximos de 0 → Pouca ou nenhuma relação linear

Através do método corr() aplicado junto a colunas numéricas.

```python
import pandas as pd

dados = {
    'Pessoa': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Idade': [23, 25, 30, 35, 40, 45, 50, 55, 60, 65],
    'Renda': [2200, 2500, 2900, 3300, 3100, 4100, 3900, 4700, 5200, 4800],
    'Gastos Mensais': [
        1500, 2000, 2200, 2600, 2400, 3100, 2700, 3500, 3300, 3700
    ]
}

df = pd.DataFrame(dados)
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cálculo da correlação/Pearson
corr_matrix = df.corr(numeric_only=True)

# Plotagem do mapa de calor com a matriz de correlação
sns.heatmap(
    corr_matrix, cmap='coolwarm', annot=True, fmt='.2f',
    vmin=-1, vmax=1, linewidths=0.6, center=0
)
plt.title("Matriz de Correlação (Pearson)")
plt.show()
```

O mapa de calor apresenta a matriz de correlação de Pearson entre as variáveis
numéricas do DataFrame. Os valores variam de -1 a 1, indicando a direção e a
intensidade da relação linear entre as variáveis.

A correlação entre Idade e Renda foi de 0,96, entre Idade e Gastos_Mensais foi
de 0,94 e entre Renda e Gastos_Mensais foi de 0,96. Esses resultados indicam
associações lineares positivas e fortes neste conjunto de dados.

No entanto, correlação não implica causalidade. Portanto, não é possível
afirmar apenas com esse gráfico que idade causa aumento da renda ou que renda
causa aumento dos gastos. É necessário analisar o contexto, a origem dos
dados, o tamanho da amostra e outros possíveis fatore

### Coeficiente de Correlação de Spearman

Mede a correlação entre variáveis quando:

- Quando pelo menos uma variável é ordinal (como níveis de satisfação, escolaridade, classificações, etc.).

- Quando os dados parecem não seguir uma relação linear, mas talvez uma curva crescente ou decrescente.

- Quando há outliers que podem distorcer a correlação de Pearson.

💻 Implementação no Pandas: Através do método **corr()** , com o parâmetro e valor **method='spearman'** , aplicado junto a colunas numéricas.

```python
import pandas as pd
import numpy as np

# Pesquisa de satisfação em uma empresa
dados = {
    'Pessoa': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Idade': [20, 25, 30, 35, 40, 45, 50, 55, 60, 65],
    # Avaliação do funcinário s/ felicidade no trabalho
    'Satisfacao': [1, 2, 3, 4, 5, 4, 3, 2, 1, 1],
    # Notas dos gestores para desempenho desse funcionário
    'Desempenho': [50, 55, 60, 70, 85, 80, 70, 60, 55, 50]
}

df = pd.DataFrame(dados)
print(df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cálculo da correlação/Spearman
corr_spearman = df.corr(method="spearman", numeric_only=True)

# Plotagem do mapa de calor com a matriz de correlação
sns.heatmap(
    corr_spearman, cmap='coolwarm', annot=True, fmt='.2f',
    vmin=-1, vmax=1, linewidths=0.6, center=0
)
plt.title("Matriz de Correlação - Spearman")
plt.show()
```

A correlação de Spearman mede relações monotônicas entre variáveis, sendo
calculada a partir dos ranks (posições) dos valores, e não dos valores brutos.
É indicada para variáveis ordinais, relações não lineares ou quando há
outliers que poderiam distorcer a correlação de Pearson.

Nos dados analisados:
- Idade x Satisfação: -0.22 (relação fraca, satisfação não sobe nem desce
  de forma consistente com a idade)
- Idade x Desempenho: -0.01 (praticamente nenhuma relação)
- Satisfação x Desempenho: 0.95 (relação forte e positiva: funcionários mais
  satisfeitos tendem a ter melhor desempenho)

Assim como na correlação de Pearson, correlação não implica causalidade.
