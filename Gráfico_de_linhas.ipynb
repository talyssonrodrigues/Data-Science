# Gráfico de linhas

Gráfico de linhas é um tipo de gráfico que mostra a informação em forma de um conjunto de pontos conectados por um segmeto de linha reta.

Esses gráficos são ideais par mostrar alterações e tendências de dados em intervalos de tempo iguais.

## *Dataset*

Para o estudo do gráfico de linhas, o dataset utilizado nesse notebook, contém o numero de casos mensais de dengue para a cidade de campinas-SP.

Além do número de casos confirmados de dengue, o dataset mostra ainda dados relacionados à chuva, temperaturas média, mínima e máxima do município. O período contemplado pelo conjunto de dados é de 1998 a 2015.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = 'https://raw.githubusercontent.com/carlosfab/dsnp2/master/datasets/dengue-dataset.csv'

df = pd.read_csv(data, index_col="data", parse_dates=True)

df.head()
```

Para plotar um gráfico com o número de casos confirmados:

```python
fig, ax = plt.subplots()

df["casos-confirmados"].plot(kind="line", ax=ax)
ax.set_title('Dengue em Campinas')
ax.set_ylabel('Casos confirmados')
ax.set_xlabel('Data')

plt.tight_layout()
```

## Selecionando intervalos

```python
fig, ax = plt.subplots()

df.loc['2011-01-01':'2013-12-31', 'temperatura-media'].plot(kind='line', ax=ax)
ax.set_title('Dengue em Campinas')
ax.set_ylabel('Casos confirmados')
ax.set_xlabel('Data')

plt.tight_layout()
```

## Plotar múltiplos gráficos separadamente

```python
fig, ax = plt.subplots(nrows = 2, ncols = 1, figsize = (10,4))

df.loc['2010-01-01':'2013-12-31', 'temperatura-media'].plot(kind='line', ax=ax[0])
ax[0].set_title('Dengue em Campinas')
ax[0].set_ylabel('temperatura média')
ax[0].set_xlabel('Data')

df.loc['2010-01-01':'2013-12-31':, 'casos-confirmados'].plot(kind='line', ax=ax[1])
ax[1].set_title('Dengue em Campinas')
ax[1].set_ylabel('Casos confirmados')
ax[1].set_xlabel('Data')

plt.tight_layout()
```

## Gráficos sobrepostos

```python
df_norm = (df - df.mean()) / df.std()
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(8,4))

df_norm.loc['2010-01-01':'2013-12-31',"temperatura-media"].plot(kind="line",
                                                                ax=ax)
df_norm.loc['2010-01-01':'2013-12-31':,"casos-confirmados"].plot(kind="line",
                                                                 ax=ax)

ax.set_xlabel("Data")
ax.set_title("Temperatura Média X Casos Confirmados")
ax.legend(["Temperatura Média", "Casos Confirmados"],loc="upper left")

plt.tight_layout()
```

Com a legenda fora:

```python
df_norm = (df - df.mean()) / df.std()
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(8,4))

df_norm.loc['2010-01-01':'2013-12-31',"temperatura-media"].plot(kind="line",
                                                                ax=ax)
df_norm.loc['2010-01-01':'2013-12-31':,"casos-confirmados"].plot(kind="line",
                                                                 ax=ax)

ax.set_xlabel("Data")
ax.set_title("Temperatura Média X Casos Confirmados")
ax.legend(["Temperatura Média", "Casos Confirmados"],
          loc="upper left",
          bbox_to_anchor=(1, 1))

plt.tight_layout()
```
