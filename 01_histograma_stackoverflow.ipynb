```python
DATA_URL = 'https://s3.amazonaws.com/video.udacity-data.com/topher/2018/February/5a8cb654_survey-results-public/survey-results-public.csv'
```

*Determinando a função get_survey_data*

```python
def get_survey_data(data_url=DATA_URL):
    csv_file = data_url.split('/')[-1]

    if not os.path.exists(csv_file):
      urllib.request.urlretrieve(DATA_URL, 'survey-results-public.csv')
      print('[+] Arquivo baixado...')
    else:
      print(f'[+] o arquivo {csv_file} já existe na pasta')
```

*Chamando a função e organizando em um data frame*

```python
get_survey_data()
```

```python
df = pd.read_csv('survey-results-public.csv')
```

```python
df.head()
```

Histograma :

```python
df['Salary'].hist();
```

**Resultado da exploração incial:**

É possível observar uma maior frequência de salários na faixa de US$ 50 mil
entre os desenvolvedores que responderam ao questionário em 2018.

A distribuição não é simétrica, pois a maior parte dos salários está
concentrada no lado esquerdo do histograma, enquanto uma quantidade menor de
observações se estende para faixas salariais mais elevadas.

Com base apenas na visualização do histograma, parece que a maior parte dos
salários está aproximadamente entre US$ 25 mil e US$ 60 mil.
